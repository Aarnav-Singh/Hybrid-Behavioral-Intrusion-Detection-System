# HB-IDS v2: End-to-End Demo
This notebook demonstrates standard data ingestion, feature generation, model forward passing, and attack simulation evaluating robustness.

```python
import pandas as pd
from backend.core.feature_engineering import FeatureEngineer
from backend.adversarial.run_attacks import run_redteam_scenario
from backend.evaluation.metrics import generate_report

# 1. Load Baseline Data
df = pd.read_csv('../examples/synthetic_dataset.csv')
```

```python
# 2. Apply Red-Team Perturbations (e.g. Feature Evasion)
target_ip = df['src_ip'].iloc[0]
poisoned_df = run_redteam_scenario(df, 'feature_perturbation', target_ip=target_ip, intensity=1.0)
poisoned_df.head()
```

```python
# 3. Feature Engineering Windowing
engineer = FeatureEngineer(window_size='5T')
windowed_df = engineer.apply_windowing(poisoned_df)
windowed_df.head()
```

```python
# 4. Generate Reports (Mocked Predictions)
import numpy as np
y_true = np.random.randint(0, 2, len(windowed_df))
y_pred_prob = np.random.uniform(0, 1, len(windowed_df))

report = generate_report(y_true, y_pred_prob)
print("Evaluation Report:", report)
```
